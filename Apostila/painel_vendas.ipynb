```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("vendas.csv")
df["data"] = pd.to_datetime(df["data"])


df["faturamento"] = df["quantidade"] * df["preco_unitario"]


print(df.info())
print(df.isnull().sum())
print(df.describe())


df["quantidade"] = df["quantidade"].fillna(df["quantidade"].mean())
df["preco_unitario"] = df["preco_unitario"].fillna(df["preco_unitario"].mean())
df["faturamento"] = df["quantidade"] * df["preco_unitario"]


faturamento_loja = df.groupby("loja")["faturamento"].sum().sort_values(ascending=False)
print(faturamento_loja)


ticket_medio = df.groupby("loja")["faturamento"].sum() / df.groupby("loja").size()
print(ticket_medio)

produtos_vendidos = df.groupby("produto")["quantidade"].sum().sort_values(ascending=False)
print(produtos_vendidos.head(5))


faturamento_mensal = df.groupby(df["data"].dt.to_period("M"))["faturamento"].sum()
print("Maior:", faturamento_mensal.idxmax(), faturamento_mensal.max())
print("Menor:", faturamento_mensal.idxmin(), faturamento_mensal.min())

df["mes"] = df["data"].dt.month


plt.figure(figsize=(8,5))
sns.barplot(x=faturamento_loja.index, y=faturamento_loja.values)
plt.title("Faturamento por Loja")
plt.xlabel("Loja")
plt.ylabel("Faturamento (R$)")
plt.show()


plt.figure(figsize=(10,5))
plt.plot(faturamento_mensal.index.astype(str), faturamento_mensal.values, marker="o")
plt.title("Evolução do Faturamento Mensal")
plt.xlabel("Mês/Ano")
plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=45)
plt.show()

faturamento_regiao = df.groupby("regiao")["faturamento"].sum()
plt.figure(figsize=(7,7))
plt.pie(faturamento_regiao.values, labels=faturamento_regiao.index, autopct="%1.1f%%")
plt.title("Faturamento por Região")
plt.show()


plt.figure(figsize=(9,5))
sns.scatterplot(data=df, x="quantidade", y="faturamento", hue="produto")
plt.title("Relação entre Quantidade e Faturamento")
plt.xlabel("Quantidade")
plt.ylabel("Faturamento (R$)")
plt.show()


plt.figure(figsize=(9,5))
sns.boxplot(data=df, x="categoria", y="preco_unitario")
plt.title("Distribuição dos Preços por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Preço Unitário (R$)")
plt.show()


correlacao = df[["quantidade", "preco_unitario", "faturamento"]].corr()
plt.figure(figsize=(7,5))
sns.heatmap(correlacao, annot=True)
plt.title("Matriz de Correlação")
plt.show()


plt.figure(figsize=(8,5))
sns.barplot(x=faturamento_loja.index, y=faturamento_loja.values)
plt.title("Faturamento Total por Loja")
plt.xlabel("Lojas")
plt.ylabel("Faturamento (R$)")
plt.show()


fig, axes = plt.subplots(1, 2, figsize=(14,5))
sns.barplot(x=faturamento_loja.index, y=faturamento_loja.values, ax=axes[0])
axes[0].set_title("Faturamento por Loja")
sns.barplot(x=faturamento_regiao.index, y=faturamento_regiao.values, ax=axes[1])
axes[1].set_title("Faturamento por Região")
plt.tight_layout()
plt.show()


resumo = pd.DataFrame({
    "indicador": [
        "Loja com maior faturamento",
        "Loja com menor faturamento",
        "Produto mais vendido",
        "Mês de maior faturamento",
        "Mês de menor faturamento"
    ],
    "resultado": [
        faturamento_loja.idxmax(),
        faturamento_loja.idxmin(),
        produtos_vendidos.idxmax(),
        str(faturamento_mensal.idxmax()),
        str(faturamento_mensal.idxmin())
    ]
})
print(resumo)
```
