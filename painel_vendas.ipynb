```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os



dados = {
    "data": [
        "2026-01-05", "2026-01-12", "2026-01-20",
        "2026-02-03", "2026-02-14", "2026-02-25",
        "2026-03-02", "2026-03-15", "2026-03-28",
        "2026-04-06", "2026-04-18", "2026-04-26",
        "2026-05-04", "2026-05-16", "2026-05-27",
        "2026-06-08", "2026-06-19", "2026-06-28",
        "2026-07-07", "2026-07-18", "2026-07-29",
        "2026-08-05", "2026-08-17", "2026-08-28",
        "2026-09-06", "2026-09-15", "2026-09-27"
    ],

    "loja": [
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C",
        "Loja A", "Loja B", "Loja C"
    ],

    "produto": [
        "Notebook", "Celular", "Tênis",
        "Celular", "Camiseta", "Notebook",
        "Tênis", "Notebook", "Camiseta",
        "Celular", "Tênis", "Notebook",
        "Notebook", "Celular", "Tênis",
        "Camiseta", "Notebook", "Celular",
        "Tênis", "Camiseta", "Notebook",
        "Celular", "Tênis", "Camiseta",
        "Notebook", "Celular", "Tênis"
    ],

    "categoria": [
        "Eletrônicos", "Eletrônicos", "Esportes",
        "Eletrônicos", "Vestuário", "Eletrônicos",
        "Esportes", "Eletrônicos", "Vestuário",
        "Eletrônicos", "Esportes", "Eletrônicos",
        "Eletrônicos", "Eletrônicos", "Esportes",
        "Vestuário", "Eletrônicos", "Eletrônicos",
        "Esportes", "Vestuário", "Eletrônicos",
        "Eletrônicos", "Esportes", "Vestuário",
        "Eletrônicos", "Eletrônicos", "Esportes"
    ],

    "regiao": [
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste",
        "Sudeste", "Sul", "Nordeste"
    ],

    "quantidade": [
        2, 5, 8,
        4, 10, 1,
        6, 2, 12,
        3, 7, 2,
        3, 6, 9,
        15, 3, 4,
        10, 14, 1,
        5, 11, 13,
        2, 7, 10
    ],

    "preco_unitario": [
        3500, 2200, 300,
        2200, 80, 3500,
        300, 3500, 80,
        2200, 300, 3500,
        3500, 2200, 300,
        80, 3500, 2200,
        300, 80, 3500,
        2200, 300, 80,
        3500, 2200, 300
    ]
}

df = pd.DataFrame(dados)


df.to_csv("vendas.csv", index=False)


df = pd.read_csv("vendas.csv")


df["data"] = pd.to_datetime(df["data"])

print(df.head())




df["faturamento"] = df["quantidade"] * df["preco_unitario"]

print(df.head())


print("Informações do dataset:")
print(df.info())

print("\nValores ausentes:")
print(df.isnull().sum())

print("\nResumo estatístico:")
print(df.describe())



df["quantidade"] = df["quantidade"].fillna(df["quantidade"].mean())
df["preco_unitario"] = df["preco_unitario"].fillna(df["preco_unitario"].mean())


df["faturamento"] = df["quantidade"] * df["preco_unitario"]

print("\nValores nulos depois do tratamento:")
print(df.isnull().sum())



faturamento_loja = df.groupby("loja")["faturamento"].sum()
faturamento_loja = faturamento_loja.sort_values(ascending=False)

print("\nFaturamento por loja:")
print(faturamento_loja)



ticket_medio = df.groupby("loja")["faturamento"].sum() / df.groupby("loja").size()

print("\nTicket médio por loja:")
print(ticket_medio)



produtos_vendidos = df.groupby("produto")["quantidade"].sum()
produtos_vendidos = produtos_vendidos.sort_values(ascending=False)

print("\nProdutos mais vendidos:")
print(produtos_vendidos.head(5))



faturamento_mes = df.groupby(df["data"].dt.to_period("M"))["faturamento"].sum()

maior_mes = faturamento_mes.idxmax()
menor_mes = faturamento_mes.idxmin()

print("\nMaior faturamento:")
print(maior_mes.strftime("%m/%Y"), "-", faturamento_mes.max())

print("\nMenor faturamento:")
print(menor_mes.strftime("%m/%Y"), "-", faturamento_mes.min())


df["mes"] = df["data"].dt.month

print("\nColuna mês:")
print(df[["data", "mes"]].head())



sns.set_style("whitegrid")
sns.set_palette("deep")


plt.figure(figsize=(8, 5))

sns.barplot(
    x=faturamento_loja.index,
    y=faturamento_loja.values
)

plt.title("Faturamento por Loja")
plt.xlabel("Loja")
plt.ylabel("Faturamento (R$)")
plt.legend(["Faturamento"])

plt.tight_layout()
plt.savefig("faturamento_por_loja.png")
plt.show()


faturamento_mensal = df.groupby(
    df["data"].dt.to_period("M")
)["faturamento"].sum()

faturamento_mensal.index = faturamento_mensal.index.strftime("%m/%Y")

plt.figure(figsize=(10, 5))

plt.plot(
    faturamento_mensal.index,
    faturamento_mensal.values,
    marker="o",
    label="Faturamento mensal"
)

plt.title("Evolução do Faturamento Mensal")
plt.xlabel("Mês/Ano")
plt.ylabel("Faturamento (R$)")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("faturamento_mensal.png")
plt.show()


faturamento_regiao = df.groupby("regiao")["faturamento"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    faturamento_regiao.values,
    labels=faturamento_regiao.index,
    autopct="%1.1f%%"
)

plt.title("Faturamento por Região")

plt.tight_layout()
plt.savefig("faturamento_por_regiao.png")
plt.show()



plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="quantidade",
    y="faturamento",
    hue="produto"
)

plt.title("Relação entre Quantidade e Faturamento")
plt.xlabel("Quantidade")
plt.ylabel("Faturamento (R$)")
plt.legend(title="Produto")

plt.tight_layout()
plt.savefig("quantidade_x_faturamento.png")
plt.show()


plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="categoria",
    y="preco_unitario"
)

plt.title("Distribuição dos Preços por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Preço Unitário (R$)")

plt.tight_layout()
plt.savefig("precos_por_categoria.png")
plt.show()


correlacao = df[
    ["quantidade", "preco_unitario", "faturamento"]
].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm"
)

plt.title("Matriz de Correlação")

plt.tight_layout()
plt.savefig("matriz_correlacao.png")
plt.show()



plt.figure(figsize=(8, 5))

sns.barplot(
    x=faturamento_loja.index,
    y=faturamento_loja.values
)

plt.title("Faturamento Total por Loja")
plt.xlabel("Lojas")
plt.ylabel("Faturamento (R$)")

plt.tight_layout()
plt.show()


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    x=faturamento_loja.index,
    y=faturamento_loja.values,
    ax=axes[0]
)

axes[0].set_title("Faturamento por Loja")
axes[0].set_xlabel("Loja")
axes[0].set_ylabel("Faturamento (R$)")


sns.barplot(
    x=faturamento_regiao.index,
    y=faturamento_regiao.values,
    ax=axes[1]
)

axes[1].set_title("Faturamento por Região")
axes[1].set_xlabel("Região")
axes[1].set_ylabel("Faturamento (R$)")

plt.tight_layout()
plt.savefig("comparacao_loja_regiao.png")
plt.show()



os.makedirs("graficos", exist_ok=True)


plt.figure(figsize=(8, 5))
sns.barplot(x=faturamento_loja.index, y=faturamento_loja.values)
plt.title("Faturamento por Loja")
plt.xlabel("Loja")
plt.ylabel("Faturamento (R$)")
plt.tight_layout()
plt.savefig("graficos/faturamento_por_loja.png")
plt.close()


plt.figure(figsize=(10, 5))
plt.plot(
    faturamento_mensal.index,
    faturamento_mensal.values,
    marker="o"
)
plt.title("Evolução do Faturamento Mensal")
plt.xlabel("Mês/Ano")
plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("graficos/faturamento_mensal.png")
plt.close()


plt.figure(figsize=(7, 7))
plt.pie(
    faturamento_regiao.values,
    labels=faturamento_regiao.index,
    autopct="%1.1f%%"
)
plt.title("Faturamento por Região")
plt.tight_layout()
plt.savefig("graficos/faturamento_por_regiao.png")
plt.close()


plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="quantidade",
    y="faturamento",
    hue="produto"
)
plt.title("Relação entre Quantidade e Faturamento")
plt.xlabel("Quantidade")
plt.ylabel("Faturamento (R$)")
plt.tight_layout()
plt.savefig("graficos/quantidade_x_faturamento.png")
plt.close()


plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x="categoria",
    y="preco_unitario"
)
plt.title("Distribuição dos Preços por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Preço Unitário (R$)")
plt.tight_layout()
plt.savefig("graficos/precos_por_categoria.png")
plt.close()


plt.figure(figsize=(7, 5))
sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm"
)
plt.title("Matriz de Correlação")
plt.tight_layout()
plt.savefig("graficos/matriz_correlacao.png")
plt.close()



resumo = pd.DataFrame({
    "indicador": [
        "Loja com maior faturamento",
        "Maior faturamento por loja",
        "Loja com menor faturamento",
        "Menor faturamento por loja",
        "Maior ticket médio",
        "Produto mais vendido",
        "Mês de maior faturamento",
        "Mês de menor faturamento",
        "Região com maior faturamento"
    ],

    "resultado": [
        faturamento_loja.idxmax(),
        f"R$ {faturamento_loja.max():.2f}",
        faturamento_loja.idxmin(),
        f"R$ {faturamento_loja.min():.2f}",
        f"R$ {ticket_medio.max():.2f}",
        produtos_vendidos.idxmax(),
        maior_mes.strftime("%m/%Y"),
        menor_mes.strftime("%m/%Y"),
        faturamento_regiao.idxmax()
    ]
})

resumo.to_csv(
    "graficos/resumo_analitico.csv",
    index=False
)

print("\nResumo analítico:")
print(resumo)



print("\nANÁLISE DOS RESULTADOS")
print("A Loja B apresentou o maior faturamento.")
print("A Loja C apresentou o menor faturamento.")
print("O produto com maior quantidade de vendas foi a Camiseta.")
print("O mês de maior faturamento foi 05/2026.")
print("O mês de menor faturamento foi 07/2026.")
print("Os gráficos mostram diferenças de faturamento entre lojas e regiões.")
```
