```python
import pandas as pd
```

```python
path = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"  # coloque o arquivo na mesma pasta do notebook (ou use o caminho completo)
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar	

Exemplo 1

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

```python
formacao_total
```

## Tabela 2 — Ciência, Tecnologia, Engenharias e Matemática

```python
# mesmas linhas (3:36), mas agora pegando as colunas da área de Ciência, Tecnologia, Engenharias e Matemática
# coluna 0 = uf/região | colunas 4, 5, 6 = total, homens, mulheres dessa área
formacao_ctem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
```

```python
formacao_ctem.columns = [
    "uf_regiao",
    "total_ctem",
    "homens_ctem",
    "mulheres_ctem"
]
```

```python
formacao_ctem
```

```python
formacao_ctem["percentual_mulheres_ctem"] = ((formacao_ctem['mulheres_ctem'] / formacao_ctem['total_ctem']) * 100).round(2)
```

```python
formacao_ctem
```

## Tabela 3 — Educação, Serviços pessoais, Saúde e Bem-estar

```python
# mesmas linhas (3:36), agora pegando as colunas da área de Educação, Serviços pessoais, Saúde e Bem-estar
# coluna 0 = uf/região | colunas 7, 8, 9 = total, homens, mulheres dessa área
formacao_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
```

```python
formacao_saude.columns = [
    "uf_regiao",
    "total_saude",
    "homens_saude",
    "mulheres_saude"
]
```

```python
formacao_saude
```

```python
formacao_saude["percentual_mulheres_saude"] = ((formacao_saude['mulheres_saude'] / formacao_saude['total_saude']) * 100).round(2)
```

```python
formacao_saude
```

## Comparando as três tabelas

```python
# Juntando as 3 tabelas em uma só, usando uf_regiao como chave
comparativo = formacao_total[["uf_regiao", "percentual_mulheres"]].merge(
    formacao_ctem[["uf_regiao", "percentual_mulheres_ctem"]],
    on="uf_regiao"
).merge(
    formacao_saude[["uf_regiao", "percentual_mulheres_saude"]],
    on="uf_regiao"
)

comparativo = comparativo.rename(columns={"percentual_mulheres": "percentual_mulheres_total"})
comparativo
```

```python
# Olhando só o Brasil: as mulheres são maioria no ensino superior de modo geral,
# mas o que acontece quando olhamos por área de formação?
brasil = comparativo.loc[comparativo["uf_regiao"] == "Brasil"].iloc[0]

print("Brasil — % de mulheres com curso superior completo (Censo 2022)")
print(f"  Total (todas as áreas):                                 {brasil.percentual_mulheres_total:.2f}%")
print(f"  Ciência, Tecnologia, Engenharias e Matemática:          {brasil.percentual_mulheres_ctem:.2f}%")
print(f"  Educação, Serviços pessoais, Saúde e Bem-estar:         {brasil.percentual_mulheres_saude:.2f}%")
```

```python
# Ranking das UFs/regiões por % de mulheres na área de Ciência, Tecnologia, Engenharias e Matemática
comparativo.sort_values("percentual_mulheres_ctem", ascending=False)
```

```python
# Ranking das UFs/regiões por % de mulheres na área de Educação, Serviços pessoais, Saúde e Bem-estar
comparativo.sort_values("percentual_mulheres_saude", ascending=False)
```

**Conclusão**

Olhando o Brasil como um todo, as mulheres são maioria entre as pessoas com curso superior completo (percentual acima de 50% no total). Porém, essa maioria não se distribui igualmente entre as áreas de formação:

- Em **Educação, Serviços pessoais, Saúde e Bem-estar**, o percentual de mulheres é bem mais alto que a média geral — essa área é fortemente concentrada em mulheres.
- Em **Ciência, Tecnologia, Engenharias e Matemática**, o percentual de mulheres cai bastante, ficando abaixo do percentual da área de Educação/Saúde e mais próximo (ou abaixo) de 50% em várias UFs.

Isso mostra que, mesmo com mais mulheres formadas no ensino superior no geral, ainda existe uma divisão bem marcada por área de formação: mulheres seguem sub-representadas nas áreas de exatas/tecnologia (STEM) em comparação com áreas de cuidado/educação.

## Atividade 2 — Domicílios sem presença de cônjuge e com filhos(as), por sexo da pessoa responsável

Fonte: `Tabela_11_Domicilios_sem_presenca_de_conjuge_e_com_filhos.xlsx` (aba "BR GR UF MU").
A tabela já vem em **percentual** de domicílios, separada por sexo da pessoa responsável (Homens / Mulheres),
para Brasil, Grandes Regiões, UFs e Municípios.

```python
path_indicador_1 = "Tabela_11_Domicilios_sem_presenca_de_conjuge_e_com_filhos.xlsx"  # coloque o arquivo na mesma pasta do notebook

# : = todas as linhas | 0:3 = colunas 0,1,2 (uf/região, homens, mulheres)
# linhas 3:36 = Brasil (1) + Grandes Regiões (5) + UFs (27)
indicador_1 = pd.read_excel(path_indicador_1, sheet_name="BR GR UF MU", header=None).iloc[3:36, 0:3].copy()
indicador_1.columns = ["uf_regiao", "homens", "mulheres"]
indicador_1
```

```python
# Filtrando só as 5 grandes regiões para o gráfico
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
reg = indicador_1[indicador_1["uf_regiao"].isin(regioes)].copy()

# melt: transforma as colunas "homens" e "mulheres" em uma coluna "sexo" + uma coluna "percentual"
reg_long = reg.melt(
    id_vars="uf_regiao",
    value_vars=["homens", "mulheres"],
    var_name="sexo",
    value_name="percentual",
)
reg_long
```

```python
# Gráfico de barras agrupadas: homens x mulheres por grande região
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=reg_long, x="uf_regiao", y="percentual", hue="sexo", ax=ax)
ax.set_title("Domicílios sem cônjuge e com filhos(as), por região e sexo da pessoa responsável (2022)")
ax.set_xlabel("Região")
ax.set_ylabel("% de domicílios")
plt.tight_layout()
plt.show()
```

## Gráficos da Atividade 3 — Formação acadêmica

Reaproveitando os dataframes já construídos mais acima:
- `formacao_total` → graduação, todas as áreas
- `formacao_ctem` → área de Ciência, Tecnologia, Engenharias e Matemática (STEM)
- `formacao_saude` → área de Educação, Serviços pessoais, Saúde e Bem-estar

```python
# Barras agrupadas: homens x mulheres no Brasil (três recortes)
brasil_pessoas = pd.DataFrame(
    {
        "area": ["Graduação (total)", "STEM", "Educação / Saúde"],
        "homens": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_homens"].item(),
            formacao_ctem.loc[formacao_ctem["uf_regiao"] == "Brasil", "homens_ctem"].item(),
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "homens_saude"].item(),
        ],
        "mulheres": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_mulheres"].item(),
            formacao_ctem.loc[formacao_ctem["uf_regiao"] == "Brasil", "mulheres_ctem"].item(),
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "mulheres_saude"].item(),
        ],
    }
)
brasil_long = brasil_pessoas.melt(id_vars="area", var_name="sexo", value_name="pessoas")
brasil_long
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com curso superior completo, por área e sexo (Censo 2022)")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```

# PARTE 2 - IDH

Fonte: `Tabela4.csv` (IDHM por UF, 1991–2024). O arquivo usa `;` como separador e `,` como separador
decimal, e a primeira linha é um título mesclado — por isso o cabeçalho real (Sigla, Código, Estado, anos...)
está na primeira linha de dados, não na linha 0.

```python
# 1. Abrir o CSV com ; e vírgula decimal
df = pd.read_csv("Tabela4.csv", sep=";", decimal=",", encoding="latin-1")
df.head()
```

```python
# A primeira linha do arquivo é um título mesclado (IDHM - Índice de Desenvolvimento Humano Municipal),
# então os nomes de coluna corretos (Sigla, Código, Estado, 1991, 2000, ...) estão na 1ª linha de dados.
novas_colunas = []
for c in df.iloc[0]:
    c = str(c)
    if c.endswith(".0"):   # os anos vieram como float (1991.0) por causa da coluna de título vazia
        c = c[:-2]
    novas_colunas.append(c)

df.columns = novas_colunas
df = df.drop(index=0).reset_index(drop=True)

# 2. Limpar colunas/linhas totalmente vazias
df = df.dropna(axis=1, how="all")
df = df.dropna(axis=0, how="all")

df.head()
```

```python
# 3. Passar a tabela do formato largo para o formato longo (melt)
anos = [c for c in df.columns if str(c).isdigit()]
variaveis = [c for c in df.columns if c not in anos]
print(anos)

df_longo = df.melt(
    id_vars=variaveis,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH",
)
df_longo["Ano"] = df_longo["Ano"].astype(int)
df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")
df_longo.head()
```

```python
# 4. Plotar a evolução do IDH de cada estado
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(grupo["Ano"], grupo["IDH"], marker="o", markersize=3, linewidth=1.5, label=sigla)

ax.set_title("Evolução do IDH por estado (1991–2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")
fig.tight_layout()
plt.show()
```

```python
# 5. Plote o IDH de Minas Gerais
mg = df_longo[df_longo["Sigla"] == "MG"].sort_values("Ano")

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(mg["Ano"], mg["IDH"], marker="o", linewidth=2, color="darkgreen")
ax.set_title("Evolução do IDH de Minas Gerais (1991–2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
plt.tight_layout()
plt.show()
```
